```python
from ultralytics import YOLO

model = YOLO('yolo11n.pt')

metrics = model.val(
    data='/content/drive/MyDrive/aquarium_dataset/aquarium_dataset/data.yaml'
)

print('Precision :', metrics.box.mp)
print('Recall :', metrics.box.mr)
print('mAP50 :', metrics.box.map50)
print('mAP50-95 :', metrics.box.map)
# Precision : 0.02657698596149982
# Recall : 0.0926869970987618
# mAP50 : 0.01988449241985285   0.5 기준에서 계산한 AP
# mAP50-95 : 0.011751987471109789
# 점수가 낮은 이뉴는 백본은 coco dataset(라벨 80개)
# aquarium_dataset
```
